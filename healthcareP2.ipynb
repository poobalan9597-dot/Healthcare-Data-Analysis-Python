```python

df['Discharge Date'] = pd.to_datetime(df['Discharge Date'])

monthly_patients = df.groupby(
    df['Discharge Date'].dt.to_period('M')
).size()


plt.figure(figsize=(14,6))

plt.plot(
    monthly_patients.index.astype(str),
    monthly_patients.values,
    marker='o'
)

plt.title('Monthly Patient Discharge Analysis')
plt.xlabel('Month')
plt.ylabel('Number of Patients')

plt.xticks(rotation=90)
plt.grid()
plt.tight_layout()
plt.show()
```

```python
df['Date of Admission'] = pd.to_datetime(df['Date of Admission'])

df['Month'] = df['Date of Admission'].dt.to_period('M')

monthly_gender = df.groupby(
    ['Month', 'Gender']
).size().unstack(fill_value=0)

plt.figure(figsize=(13,5))

plt.plot(
    monthly_gender.index.astype(str),
    monthly_gender['Male'],
    marker='o',
    label='Male'
)

plt.plot(
    monthly_gender.index.astype(str),
    monthly_gender['Female'],
    marker='o',
    label='Female'
)

plt.title('Monthly Admission: Male vs Female')
plt.xlabel('Month')
plt.ylabel('Number of Patients')
plt.xticks(rotation=90)
plt.legend()
plt.grid()
plt.tight_layout()
plt.show()


```
